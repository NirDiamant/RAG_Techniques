# Late Chunking

```python
from langchain_openai import OpenAIEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
doc = open('data/Understanding_Climate_Change.pdf', 'rb').read().decode('latin-1')
chunks = RecursiveCharacterTextSplitter(chunk_size=500).split_text(doc)
```

```python
vectors = OpenAIEmbeddings().embed_documents(chunks)
print(len(vectors))
```
